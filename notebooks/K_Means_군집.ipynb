```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
import pandas as pd

df = pd.read_csv("df_final_final_add2.csv", encoding="cp949")
df.head()
```

```python
df[['name', 'cleaned_tokens_add']].head()
```

```python
coffee_keywords = [
    '커피', '라떼', '원두', '드립', '에스프레소', '아메리카노', '음료',
    '로스팅', '크림', '카푸치노', '플랫화이트', '콜드브루', '핸드드립',
    '티', '차', '에이드', '주스', '밀크티', '바리스타', '로스'
]

dessert_keywords = [
    '디저트', '베이커리', '빵집', '빵', '바게트', '케이크', '에그타르트',
    '타르트', '소금빵', '소금', '샌드위치', '브런치', '초코', '쿠키',
    '휘낭시에', '마들렌', '스콘', '크루아상', '크로와상', '도넛',
    '페이스트리', '베이글', '잠봉', '치아바타', '사워도'
]

space_keywords = [
    '분위기', '감성', '데이트', '공간', '내부', '인테리어', '조용하다',
    '조용한', '자리', '좌석', '테라스', '사진', '뷰', '주말', '아늑하다',
    '아늑한', '넓다', '넓은', '예쁘다', '예쁜', '편안하다', '편안한',
    '작업', '대화', '모임', '외관', '테이블', '평일'
]
```

```python
def count_keywords(text, keywords):
    tokens = str(text).split()
    return sum(token in keywords for token in tokens)

df['커피·음료'] = df['cleaned_tokens_add'].apply(
    lambda x: count_keywords(x, coffee_keywords)
)

df['디저트·베이커리'] = df['cleaned_tokens_add'].apply(
    lambda x: count_keywords(x, dessert_keywords)
)

df['공간·분위기'] = df['cleaned_tokens_add'].apply(
    lambda x: count_keywords(x, space_keywords)
)
```

```python
df[['name', '커피·음료', '디저트·베이커리', '공간·분위기']].head()
```

```python
cafe_profile = df.groupby('name')[
    ['커피·음료', '디저트·베이커리', '공간·분위기']
].sum().reset_index()

cafe_profile.head()
```

```python
total = cafe_profile[
    ['커피·음료', '디저트·베이커리', '공간·분위기']
].sum(axis=1)

cafe_profile['커피·음료 경험'] = cafe_profile['커피·음료'] / total
cafe_profile['디저트·베이커리 경험'] = cafe_profile['디저트·베이커리'] / total
cafe_profile['공간·분위기 경험'] = cafe_profile['공간·분위기'] / total
```

```python
cafe_profile.shape
```

```python
X = cafe_profile[
    ['커피·음료 경험', '디저트·베이커리 경험', '공간·분위기 경험']
]
scores = []

for k in range(2, 7):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    scores.append([k, score])

score_df = pd.DataFrame(scores, columns=['K', 'Silhouette Score'])
score_df
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

cafe_profile['cluster'] = kmeans.fit_predict(X)
```

```python
centers = pd.DataFrame(
    kmeans.cluster_centers_,
    columns=['커피·음료 경험', '디저트·베이커리 경험', '공간·분위기 경험']
)

centers.round(3)
```

```python
cafe_profile['cluster'].value_counts().sort_index()
```

```python
cluster_names = {
    0: '베이커리 특화형',
    1: '공간·분위기 중심형',
    2: '디저트·베이커리 중심형',
    3: '커피·음료 중심형'
}

cafe_profile['cluster_name'] = cafe_profile['cluster'].map(cluster_names)

```

```python
cafe_profile.groupby('cluster_name')[
    ['커피·음료 경험', '디저트·베이커리 경험', '공간·분위기 경험']
].mean().round(3)
```

```python
result = cafe_profile[
    ['name',
     '커피·음료 경험',
     '디저트·베이커리 경험',
     '공간·분위기 경험',
     'cluster',
     'cluster_name']
]

result.to_csv(
    'cafe_kmeans_cluster_results.csv',
    index=False,
    encoding='utf-8-sig'
)

result.head()
```
